## 1. My rule and its reason codes

### Regra inicial de priorização

Minha baseline será uma regra transparente para identificar conteúdos com oportunidade de melhoria orgânica. Ela priorizará conteúdos com volume relevante de impressões, posição média disponível e CTR inferior à referência observada para conteúdos da mesma faixa de posição.

A referência de CTR será calculada a partir dos dados observados, considerando também os registros com CTR zero. Os conteúdos com CTR positivo serão analisados separadamente para compreender o desempenho entre aqueles que receberam cliques.

**Códigos de motivo:**

- `HIGH_IMPRESSIONS`: conteúdo com volume relevante de impressões.
- `LOW_CTR_FOR_POSITION`: CTR abaixo da referência definida para sua faixa de posição.
- `REVIEW_OPPORTUNITY`: conteúdo selecionado para avaliação humana antes de qualquer ação.

Esta é uma baseline baseada em regras, não um modelo treinado. Os limites serão definidos a partir da distribuição dos dados e sua qualidade será avaliada pela precisão dos itens prioritários revisados manualmente.

```python
# ============================================================
# ML-07 — BASELINE TRANSPARENTE DE PRIORIZAÇÃO
# ============================================================

import pandas as pd
import numpy as np

# 1. Preparar os dados
baseline = dados.copy()

colunas = [
    "content_id",
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "avg_position",
    "position_tier"
]

baseline = baseline[colunas].copy()

# Garantir valores numéricos
for coluna in ["impressions_90d", "clicks_90d", "ctr", "avg_position"]:
    baseline[coluna] = pd.to_numeric(
        baseline[coluna], errors="coerce"
    )

# 2. Remover registros sem dados essenciais
baseline = baseline.dropna(
    subset=["impressions_90d", "ctr", "avg_position", "position_tier"]
)

# Posição zero significa ausência de posição válida
baseline = baseline[baseline["avg_position"] > 0].copy()

# 3. Calcular CTR de referência por faixa de posição
referencias = (
    baseline[baseline["ctr"] > 0]
    .groupby("position_tier")["ctr"]
    .median()
    .rename("ctr_referencia")
)

baseline = baseline.join(
    referencias,
    on="position_tier"
)

# 4. Definir critérios transparentes
baseline["alta_exposicao"] = (
    baseline["impressions_90d"] >= 500
)

baseline["ctr_abaixo_referencia"] = (
    baseline["ctr"] < baseline["ctr_referencia"]
)

# 5. Calcular oportunidade de melhoria
baseline["gap_ctr"] = (
    baseline["ctr_referencia"] - baseline["ctr"]
).clip(lower=0)

baseline["score"] = (
    baseline["impressions_90d"] *
    baseline["gap_ctr"] /
    baseline["ctr_referencia"].replace(0, np.nan)
)

# 6. Criar códigos de motivo
baseline["reason_codes"] = baseline.apply(
    lambda linha: ", ".join([
        codigo
        for condicao, codigo in [
            (linha["alta_exposicao"], "HIGH_IMPRESSIONS"),
            (
                linha["ctr_abaixo_referencia"],
                "LOW_CTR_FOR_POSITION"
            )
        ]
        if condicao
    ]),
    axis=1
)

# 7. Selecionar oportunidades válidas
oportunidades = baseline[
    baseline["alta_exposicao"] &
    baseline["ctr_abaixo_referencia"]
].copy()

# 8. Ordenar e selecionar Top 20
top_20 = oportunidades.sort_values(
    by="score",
    ascending=False
).head(20)

print("REGISTROS ANALISADOS:", len(baseline))
print("OPORTUNIDADES IDENTIFICADAS:", len(oportunidades))
print("\nTOP 20 CONTEÚDOS PRIORITÁRIOS:")

display(
    top_20[
        [
            "content_id",
            "impressions_90d",
            "ctr",
            "ctr_referencia",
            "avg_position",
            "score",
            "reason_codes"
        ]
    ].reset_index(drop=True)
)
```

```python
# ============================================================
# AVALIAÇÃO DA BASELINE — PRECISION@20
# ============================================================

# Verificar se existe uma coluna de referência para avaliação
coluna_alvo = "is_declining_label"

if coluna_alvo in dados.columns:

    # Associar o rótulo real aos conteúdos selecionados
    avaliacao = top_20.merge(
        dados[["content_id", coluna_alvo]],
        on="content_id",
        how="left"
    )

    print("AVALIAÇÃO DA BASELINE")

    print("\nDistribuição dos rótulos no Top 20:")
    display(avaliacao[coluna_alvo].value_counts(dropna=False))

    # Calcular Precision@20
    precision_20 = avaliacao[coluna_alvo].mean()

    print(
        "\nPRECISION@20:",
        round(precision_20, 4)
    )

else:
    print(
        "A coluna is_declining_label não foi encontrada."
    )
    print("Colunas disponíveis:")
    print(dados.columns.tolist())
```

```python
# ============================================================
# AVALIAÇÃO DA BASELINE — PRECISION@20
# ============================================================

# Recuperar o rótulo de tendência dos conteúdos selecionados
avaliacao = top_20.merge(
    dados[["content_id", "trend_direction"]],
    on="content_id",
    how="left"
)

# Considerar "down" como resultado positivo
avaliacao["resultado_real"] = (
    avaliacao["trend_direction"] == "down"
)

# Contabilizar os resultados
positivos = avaliacao["resultado_real"].sum()
total = len(avaliacao)

precision_20 = positivos / total if total > 0 else 0

print("AVALIAÇÃO DA BASELINE")
print("-----------------------------")
print("Conteúdos avaliados:", total)
print("Conteúdos em declínio:", positivos)
print("Precision@20:", round(precision_20, 4))
print("Precision@20 (%):", round(precision_20 * 100, 2), "%")

print("\nDISTRIBUIÇÃO DAS TENDÊNCIAS NO TOP 20:")
display(
    avaliacao["trend_direction"].value_counts(dropna=False)
)

print("\nREVISÃO DOS 20 CONTEÚDOS:")
display(
    avaliacao[
        [
            "content_id",
            "impressions_90d",
            "ctr",
            "ctr_referencia",
            "score",
            "trend_direction"
        ]
    ]
)
```

```python
# ============================================================
# COMPARAÇÃO DA BASELINE COM A DISTRIBUIÇÃO GERAL
# ============================================================

# Proporção geral de conteúdos em declínio
base_valid = dados["trend_direction"].dropna()

prevalencia_down = (
    base_valid == "down"
).mean()

# Precision@20 já calculada anteriormente
precision_baseline = precision_20

# Lift da baseline em relação à proporção geral
lift = (
    precision_baseline / prevalencia_down
    if prevalencia_down > 0 else 0
)

print("COMPARAÇÃO DA BASELINE")
print("------------------------------")
print("Total de registros válidos:", len(base_valid))
print("Proporção geral de conteúdos down:",
      round(prevalencia_down * 100, 2), "%")
print("Precision@20 da baseline:",
      round(precision_baseline * 100, 2), "%")
print("Lift:", round(lift, 3))

if lift > 1:
    print("\nA baseline concentra mais conteúdos down que a média geral.")
elif lift < 1:
    print("\nA baseline concentra menos conteúdos down que a média geral.")
else:
    print("\nA baseline apresenta resultado igual à média geral.")
```

```python
import numpy as np
from pathlib import Path

# 1. Validar as colunas necessárias
colunas_necessarias = [
    "content_id",
    "impressions_90d",
    "ctr",
    "ctr_referencia",
    "avg_position"
]

faltantes = [
    coluna for coluna in colunas_necessarias
    if coluna not in analise.columns
]

if faltantes:
    raise ValueError(f"Colunas não encontradas: {faltantes}")

# 2. Construir o ranking
resultado = analise.copy()

resultado["score"] = (
    resultado["impressions_90d"]
    * (1 - resultado["ctr"] / resultado["ctr_referencia"])
).clip(lower=0)

# 3. Criar códigos de motivo
limite_impressoes = resultado["impressions_90d"].quantile(0.75)

def identificar_motivos(linha):
    motivos = []

    if linha["impressions_90d"] >= limite_impressoes:
        motivos.append("HIGH_IMPRESSIONS")

    if linha["ctr"] < linha["ctr_referencia"]:
        motivos.append("LOW_CTR_FOR_POSITION")

    return ", ".join(motivos) if motivos else "NO_PRIORITY_SIGNAL"

resultado["reason_codes"] = resultado.apply(
    identificar_motivos, axis=1
)

# 4. Ordenar os conteúdos por pontuação
resultado = resultado.sort_values(
    by="score",
    ascending=False
).reset_index(drop=True)

# 5. Salvar o CSV
pasta_saida = Path("work/outputs")
pasta_saida.mkdir(parents=True, exist_ok=True)

caminho_saida = pasta_saida / "baseline_action_score.csv"

resultado.to_csv(caminho_saida, index=False)

# 6. Conferir o resultado
print("RANKING CONSTRUÍDO COM SUCESSO")
print("Total de conteúdos:", len(resultado))
print("Arquivo gerado:", caminho_saida)

print("\nTOP 10 CONTEÚDOS PRIORITÁRIOS:")
display(resultado.head(10))

print("\nDISTRIBUIÇÃO DOS MOTIVOS:")
display(
    resultado["reason_codes"]
    .value_counts()
    .to_frame("quantidade")
)
```

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

```python

def avaliar_conteudo(row):

    # 1. Capturar os indicadores do conteúdo
    tendencia = row.get("trend_direction", "unknown")

    impressoes = pd.to_numeric(
        row.get("impressions_90d"), errors="coerce"
    )

    ctr = pd.to_numeric(
        row.get("ctr"), errors="coerce"
    )

    ctr_referencia = pd.to_numeric(
        row.get("ctr_referencia"), errors="coerce"
    )

    # 2. Definir ação e urgência conforme a tendência
    if tendencia == "down":
        acao = (
            "Priorizar revisão do título, descrição "
            "e alinhamento com a intenção de busca."
        )
        urgencia = "Alta"

    elif tendencia == "stable":
        acao = (
            "Revisar título, descrição e alinhamento "
            "com a intenção de busca."
        )
        urgencia = "Média"

    else:
        acao = (
            "Investigar a evolução positiva antes "
            "de realizar alterações."
        )
        urgencia = "Monitoramento"

    # 3. Avaliar o volume de exposição
    if pd.notna(impressoes):

        if impressoes >= 500:
            nivel_exposicao = "elevado"

        elif impressoes >= 100:
            nivel_exposicao = "intermediário"

        else:
            nivel_exposicao = "limitado"

        texto_impressoes = f"{impressoes:,.0f}"

    else:
        nivel_exposicao = "não disponível"
        texto_impressoes = "não informado"

    # 4. Comparar CTR observado com a referência
    if pd.notna(ctr) and pd.notna(ctr_referencia):

        diferenca_ctr = ctr_referencia - ctr

        texto_ctr = (
            f"CTR observado de {ctr:.1%}, "
            f"referência de {ctr_referencia:.1%}, "
            f"diferença de {diferenca_ctr:.1%}"
        )

        if diferenca_ctr > 0:
            diagnostico = "abaixo da referência"

        elif diferenca_ctr < 0:
            diagnostico = "acima da referência"

        else:
            diagnostico = "igual à referência"

    else:
        diferenca_ctr = None
        texto_ctr = "CTR ou referência não disponível"
        diagnostico = "comparação indisponível"

    # 5. Construir nota de confiança individualizada
    confianca = (
        f"Evidência direcional baseada em {texto_impressoes} "
        f"impressões nos últimos 90 dias. "
        f"{texto_ctr}. "
        f"A confiança é limitada porque os indicadores "
        f"não comprovam a causa do desempenho."
    )

    # 6. Construir ressalva conforme os dados observados
    if tendencia == "up":
        contexto_tendencia = (
            "A tendência recente é positiva; "
            "alterações precipitadas podem interromper essa evolução."
        )

    elif tendencia == "down":
        contexto_tendencia = (
            "A tendência recente é negativa, mas não identifica "
            "sozinha a causa da queda."
        )

    else:
        contexto_tendencia = (
            "A tendência estável não garante que o desempenho "
            "permanecerá constante."
        )

    ressalva = (
        f"{contexto_tendencia} "
        f"A comparação também pode estar equivocada se a referência "
        f"de CTR não representar adequadamente a posição e a intenção "
        f"de busca, ou se houver sazonalidade, mudanças na SERP "
        f"ou problemas de mensuração."
    )

    # 7. Retornar avaliação estruturada
    return pd.Series({
        "action": acao,
        "urgency": urgencia,
        "reason_code": row["reason_codes"],
        "confidence_note": confianca,
        "what_would_make_it_wrong": ressalva
    })
```

## 4. Weak picks + leakage check

A análise inicial indica que a pontuação do ranking utiliza as variáveis `impressions_90d`, `ctr` e `ctr_referencia`. As variáveis `trend_direction`, `trend_pct` e `days_since_last_update` não aparecem diretamente na fórmula de pontuação inspecionada.

A inspeção dos nomes das colunas não identificou variáveis explicitamente relacionadas a produtos. Essa verificação é inicial e não permite descartar possíveis relações indiretas.

Também foi identificada a variável temporal `days_since_last_update`. Sua presença, isoladamente, não caracteriza vazamento de dados. É necessário verificar como as variáveis foram construídas e se as informações estavam disponíveis no momento da decisão.
